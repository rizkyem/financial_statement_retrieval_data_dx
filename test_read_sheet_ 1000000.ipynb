```python
import pandas as pd
import openpyxl

df = pd.read_excel(
    "financial_statement/2025/Audit/FinancialStatement-2025-Tahunan-BBCA.xlsx",
    sheet_name="1000000",
    header=None
)

data = {}

for i in range(4, len(df)):
    row = df.iloc[i]
    col_name = row[2]
    value    = row[1]
    if pd.notna(col_name):
        data[col_name] = value

result = pd.DataFrame([data])

# Reorder: Entity code first, then Entity name, then the rest
priority = ["Entity code", "Entity name"]
front    = [c for c in priority if c in result.columns]
rest     = [c for c in result.columns if c not in priority]
result   = result[front + rest]

result
```
